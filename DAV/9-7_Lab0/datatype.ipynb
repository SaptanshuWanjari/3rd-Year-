# Conditional Statements

```python
x=10
if x>5:
    print(f"{x} is greater than 5")

y=3
if y%2==0:
    print(f"{y} is even")
else:
    print(f"{y} is odd")

score=85
if score>=90:
    print(f"Grade:{score}\nA")
elif score>=80:
    print(f"Grade:{score}\nB")
elif score>=70:
    print(f"Grade: {score}\nC")
else:
    print(f"Grade: {score}\nD")
```

```python
secret=7
guess=0
while True:
    guess+=1
    if guess==secret:
        print("Congratulations! You guessed the secret number.")
        break
    elif guess>secret+2:
        print("Too many wrong guesses!")
        break
    else:
        print(f"Guess: {guess}: Not correct...")
```

```python
def power(base,exp=2):
    return base**exp

print( power(2) )
print( power(2,5) )
```

```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)

factorial(5)
```

```python
def fibonacci(n):
    if n<=0:
        return []
    if n==1:
        return [0]
    if n==2:
        return [0,1]
    else:
        fib=fibonacci(n-1)
        fib.append(fib[-1]+fib[-2])
        return fib

fibonacci(10)
```
